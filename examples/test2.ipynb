```python
import os
import csv

data_dir = "/tmp/data"
os.makedirs(data_dir, exist_ok=True)
file_path = os.path.join(data_dir, "temperatures.csv")
temperatures = [12.5, 15.5]
try:
    with open(file_path, "w", newline="") as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(["temperature"])
        for temp in temperatures:
            writer.writerow([temp])
except Exception as e:
    raise RuntimeError(f"Failed to write CSV file at {file_path}: {e}")
    
print(file_path)
```

## Calculate mean temperature

Read `/tmp/data/temperatures.csv`, calculate the arithmetic mean of the
`temperature` column, and write a one-row CSV file to
`/tmp/data/mean_temperature.csv`.

```python
import os
import csv

# Create directory if it doesn't exist
os.makedirs('/tmp/data', exist_ok=True)

# Read input file
input_path = '/tmp/data/temperatures.csv'
output_path = '/tmp/data/mean_temperature.csv'

# Initialize variables
total = 0.0
count = 0

try:
    with open(input_path, 'r') as f:
        reader = csv.DictReader(f)
        for row in reader:
            try:
                temp = float(row['temperature'])
                total += temp
                count += 1
            except (ValueError, KeyError) as e:
                raise ValueError(f"Invalid temperature value or missing column: {e}")
                
    if count == 0:
        raise ValueError("No valid temperature data found in input file")
        
    mean_temp = total / count

except FileNotFoundError:
    raise FileNotFoundError(f"Input file not found: {input_path}")

# Write output
with open(output_path, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['mean_temperature'])
    writer.writerow([mean_temp])
```
